```python
# ============================================================================
# RESET MY DATABASE (safe) -- wipes only YOUR database, not the whole instance
# ============================================================================
# This deletes every node/relationship and drops any leftover GDS graph
# projections inside YOUR database only (the "database" variable above). It
# does NOT drop/recreate the database itself (no admin rights needed), and it
# never touches student1..N databases belonging to anyone else. Safe to run
# any time you want a clean slate; safe to re-run.

CONFIRM_RESET = True # flip to True, then run this cell, to actually reset YOUR database

def reset_my_database():
    with get_driver().session(database=database) as session:
        try:
            projections = session.run(
                "CALL gds.graph.list() YIELD graphName RETURN graphName"
            ).data()
            for p in projections:
                session.run("CALL gds.graph.drop($name, false)", {"name": p["graphName"]})
        except Exception as e:
            print(f"(skipped GDS projection cleanup: {e})")
        session.run("MATCH (n) DETACH DELETE n")
    print(f"[OK] Database '{database}' has been reset. Other students' databases are untouched.")

if CONFIRM_RESET:
    reset_my_database()
else:
    print("Skipped -- set CONFIRM_RESET = True and re-run this cell to actually reset your database.")
```

```python
import pandas as pd

machines = pd.DataFrame({
    "MACHINE_ID": list(range(1, 21)),
    "machine_type": [
        "Cutter", "Welder", "Press", "Assembler", "Paint", "Cutter", "Welder", "Press", "Assembler", "Paint",
        "Cutter", "Welder", "Press", "Assembler", "Paint",
        "Cutter", "Welder", "Press", "Assembler", "Assembler",
    ],
    "current_status": ["active"] * 20,
    "risk_level": [
        "low", "low", "medium", "medium", "low",
        "low", "medium", "medium", "low", "low",
        "medium", "high", "medium", "high", "medium",
        "low", "medium", "low", "medium", "high",
    ],
})

sensors = pd.DataFrame({
    "SENSOR_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "sensor_type": (["temperature"] * 5) + (["vibration"] * 5),
})

# Feed_Relationships.csv, verbatim -- 27 raw rows. Rows 10 and 12 (0-indexed: the
# second "5,6" and second "8,9") are duplicates of an existing edge with a new
# throughput reading; rows 11 and 13 ("6,5" and "9,8") are the *reverse* of an
# edge already present -- MERGE below turns those two reversed rows into the
# small rework loops the SCC step finds.
feed_relationships = pd.DataFrame({
    "SRC_MACHINE_ID": [1, 2, 3, 4, 5, 6, 7, 8, 9, 5, 6, 8, 9, 1, 2, 3, 4,
                        11, 12, 13, 14, 15, 16, 17, 18, 19, 3],
    "DST_MACHINE_ID": [2, 3, 4, 5, 6, 7, 8, 9, 10, 6, 5, 9, 8, 20, 20, 20, 20,
                        12, 13, 14, 15, 16, 17, 18, 19, 20, 11],
    "THROUGHPUT_RATE": [50, 50, 50, 50, 50, 50, 50, 50, 50, 40, 40, 30, 30,
                         200, 180, 160, 140, 20, 20, 20, 20, 20, 20, 20, 20, 20, 120],
})

# Sensor_Logs.csv, verbatim -- 16 readings. Only sensors 101, 102, 106 and 107
# ever appear here; sensors 103-105 and 108-110 were registered in Sensors.csv
# but never wired to a machine -- WCC will surface those as orphans later.
sensor_logs = pd.DataFrame({
    "SRC_SENSOR_ID": [101, 101, 101, 102, 102, 102, 106, 106, 106, 107, 107, 107,
                       101, 102, 106, 107],
    "DST_MACHINE_ID": [1, 1, 1, 2, 2, 2, 6, 6, 6, 7, 7, 7, 1, 2, 6, 7],
    "READING_TYPE": ["temperature", "temperature", "temperature",
                      "temperature", "temperature", "temperature",
                      "vibration", "vibration", "vibration",
                      "vibration", "vibration", "vibration",
                      "temperature", "temperature", "vibration", "vibration"],
    "READING_VALUE": [98.5, 101.2, 99.8, 100.4, 97.9, 102.3,
                       0.03, 0.05, 0.04, 0.02, 0.06, 0.03,
                       200.0, 5.0, 0.01, 0.10],  # last 4 rows are anomaly spikes/drops
    "TS": ["2025-04-27T10:00:00Z", "2025-04-27T11:00:00Z", "2025-04-27T12:00:00Z",
           "2025-04-27T10:15:00Z", "2025-04-27T11:15:00Z", "2025-04-27T12:15:00Z",
           "2025-04-27T10:30:00Z", "2025-04-27T11:30:00Z", "2025-04-27T12:30:00Z",
           "2025-04-27T10:45:00Z", "2025-04-27T11:45:00Z", "2025-04-27T12:45:00Z",
           "2025-04-27T14:00:00Z", "2025-04-27T15:00:00Z", "2025-04-27T14:30:00Z", "2025-04-27T15:30:00Z"],
})

print("[OK] Real plant data loaded!")
print("Machines:", machines.shape, "Sensors:", sensors.shape)
print("Feed relationship rows (raw, incl. duplicates/reverse edges):", feed_relationships.shape)
print("Sensor log rows:", sensor_logs.shape)
```

```python
constraints = [
    "CREATE CONSTRAINT machine_id IF NOT EXISTS FOR (m:Machine) REQUIRE m.id IS UNIQUE",
    "CREATE CONSTRAINT sensor_id IF NOT EXISTS FOR (s:Sensor) REQUIRE s.id IS UNIQUE"
]

with get_driver().session(database=database) as session:
    for q in constraints:
        session.run(q)
print("[OK] Constraints created!")
```

```python
with get_driver().session(database=database) as session:
    for _, row in machines.iterrows():
        session.run("""
            MERGE (m:Machine {id:$MACHINE_ID})
            SET m.type = $machine_type,
                m.risk_level = $risk_level,
                m.status = $current_status
        """, row.to_dict())

    for _, row in sensors.iterrows():
        session.run("""
            MERGE (s:Sensor {id:$SENSOR_ID})
            SET s.type = $sensor_type
        """, row.to_dict())

    for _, row in feed_relationships.iterrows():
        session.run("""
            MATCH (src:Machine {id:$SRC_MACHINE_ID})
            MATCH (dst:Machine {id:$DST_MACHINE_ID})
            MERGE (src)-[r:FEEDS_INTO]->(dst)
            SET r.throughput_rate = $THROUGHPUT_RATE
        """, row.to_dict())

    for _, row in sensor_logs.iterrows():
        session.run("""
            MATCH (s:Sensor {id:$SRC_SENSOR_ID})
            MATCH (m:Machine {id:$DST_MACHINE_ID})
            MERGE (s)-[r:SENDS_READING_TO {timestamp:$TS}]->(m)
            SET r.reading_type = $READING_TYPE,
                r.value = $READING_VALUE
        """, row.to_dict())

print("[OK] Machines, sensors, feed relationships and sensor readings loaded!")
```

```python
from neo4j.graph import Path, Node, Relationship
from pyvis.network import Network
from IPython.display import IFrame

COLOR_BY_LABEL = {"Machine": "#FB923C", "Sensor": "#22D3EE"}
DEFAULT_COLOR = "#A78BFA"
NAME_PROPS = ("id", "type")

def visualize(records, height="600px", filename="graph.html"):
    net = Network(height=height, width="100%", directed=True, notebook=True,
                  cdn_resources="in_line", bgcolor="#0b1220", font_color="#f8fafc")
    net.barnes_hut(spring_length=180)
    # ... add_node()/add_relationship() walk record["path"] or record["nodes"]/["relationships"]
    # (full helper body omitted here for brevity -- walks nodes/relationships into the net object)
    net.show(filename)
    return IFrame(filename, width="100%", height=height)
```

```python
with get_driver().session(database=database) as session:
    # Quick overview of your graph data
    result = session.run("""
    MATCH (n)
    RETURN labels(n)[0] AS Label, COUNT(*) AS Count
    ORDER BY Count DESC
    """)
    node_summary = result.data()

    result = session.run("""
    MATCH ()-[r]->()
    RETURN type(r) AS Relationship, COUNT(*) AS Count
    ORDER BY Count DESC
    """)
    rel_summary = result.data()

print("[STATS] Node Summary:")
for r in node_summary:
    print(f"{r['Label']}: {r['Count']}")

print("\n[LINK] Relationship Summary:")
for r in rel_summary:
    print(f"{r['Relationship']}: {r['Count']}")
```

```python
with get_driver().session(database=database) as session:
    result = session.run("""
    MATCH (m1:Machine)-[r]->(m2:Machine)
    RETURN m1.id AS Source, type(r) AS Relationship, m2.id AS Target
    LIMIT 10
    """)
    rows = result.data()

print("[SEARCH] Sample Machine Connections:")
for r in rows:
    print(f"{r['Source']} -[{r['Relationship']}]-> {r['Target']}")
```

```python
with get_driver().session(database=database) as session:
    # Out-degree centrality (how many others each machine feeds)
    result = session.run("""
    MATCH (m:Machine)-[:FEEDS_INTO]->(x)
    RETURN m.id AS Machine, COUNT(x) AS OutDegree
    ORDER BY OutDegree DESC
    LIMIT 10
    """)
    out_rows = result.data()

    # In-degree centrality (how many feed into it)
    result = session.run("""
    MATCH (x)-[:FEEDS_INTO]->(m:Machine)
    RETURN m.id AS Machine, COUNT(x) AS InDegree
    ORDER BY InDegree DESC
    LIMIT 10
    """)
    in_rows = result.data()

print("[FACTORY] Machines with highest out-degree (producers):")
for r in out_rows:
    print(f"Machine {r['Machine']} -> {r['OutDegree']} outputs")

print("\n[RECEIVER] Machines with highest in-degree (receivers):")
for r in in_rows:
    print(f"Machine {r['Machine']} <- {r['InDegree']} inputs")
```

```python
query = """
MATCH (m:Machine)
RETURN m.id AS machine, m.wcc AS component
ORDER BY component, machine
"""
with get_driver().session(database=database) as session:
    wcc_df = pd.DataFrame(session.run(query).data())
wcc_df.groupby("component")["machine"].count().reset_index(name="machine_count")
```

```python
query = """
MATCH (n)
RETURN labels(n)[0] AS label, n.id AS id, n.wcc AS component
ORDER BY component, label, id
"""
with get_driver().session(database=database) as session:
    full_wcc_df = pd.DataFrame(session.run(query).data())

component_sizes = full_wcc_df.groupby("component").size().reset_index(name="node_count")
print("[STATS] Component sizes on the FULL graph (machines + sensors):")
print(component_sizes.to_string(index=False))

singleton_components = component_sizes[component_sizes["node_count"] == 1]["component"]
orphans = full_wcc_df[full_wcc_df["component"].isin(singleton_components)]
print("\n[WARNING] Orphan nodes (their own size-1 component -- never wired to the rest of the plant):")
print(orphans[["label", "id"]].to_string(index=False))
```

```python
with get_driver().session(database=database) as session:
    result = session.run("""
    CALL gds.degree.stream('full', {
        orientation: 'REVERSE',
        nodeLabels: ['Machine'],
        relationshipTypes: ['FEEDS_INTO']
    })
    YIELD nodeId, score
    RETURN gds.util.asNode(nodeId).id AS machine, score
    ORDER BY score DESC
    LIMIT 10
    """)
    rows = [{"Machine": r["machine"], "InDegree": int(r["score"])} for r in result]

print("[NAIVE] Top 10 Machines by in-degree (naive importance):")
for r in rows:
    print(f"Machine {r['Machine']} <- {r['InDegree']} inputs")
```

```python
with get_driver().session(database=database) as session:
    result = session.run("""
    CALL gds.pageRank.stream('full', {
        maxIterations: 20,
        dampingFactor: 0.85,
        nodeLabels: ['Machine'],
        relationshipTypes: ['FEEDS_INTO']
    })
    YIELD nodeId, score
    RETURN gds.util.asNode(nodeId).id AS machine, score
    ORDER BY score DESC
    LIMIT 10
    """)

    # Collect results
    rows = [{"Machine": r["machine"], "Score": round(r["score"], 4)} for r in result]

# Print as a formatted table
if rows:
    print("[FACTORY] Top 10 Machines by PageRank:\n")
    # Determine column widths
    col1_width = max(len(str(r["Machine"])) for r in rows + [{"Machine": "Machine"}])
    col2_width = max(len(str(r["Score"])) for r in rows + [{"Score": "Score"}])
    header = f"| {'Machine'.ljust(col1_width)} | {'Score'.ljust(col2_width)} |"
    divider = f"|{'-'*(col1_width+2)}|{'-'*(col2_width+2)}|"
    print(header)
    print(divider)
    for r in rows:
        print(f"| {str(r['Machine']).ljust(col1_width)} | {str(r['Score']).ljust(col2_width)} |")
else:
    print("No results found.")
```

```python
with get_driver().session(database=database) as session:
    result = session.run("""
    CALL gds.knn.stream('embeddings', { topK: 5, nodeProperties: ['embedding'] })
    YIELD node1, node2, similarity
    RETURN gds.util.asNode(node1).id AS machine1,
           gds.util.asNode(node2).id AS machine2,
           similarity
    ORDER BY similarity DESC
    LIMIT 10
    """)
    rows = result.data()

print("[SEARCH] Most structurally similar machine pairs:")
for r in rows:
    print(f"{r['machine1']} <-> {r['machine2']} (similarity={r['similarity']:.3f})")
```
